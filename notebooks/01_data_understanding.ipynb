```python
# Convert transaction time from seconds to hours

df_clean["Time_hours"] = df_clean["Time"] / 3600

# Display results

print(df_clean[["Time", "Time_hours"]].head(10))

# Check time range

print("\nMinimum Time (hours):", df_clean["Time_hours"].min())

print("Maximum Time (hours):", df_clean["Time_hours"].max())
```

```python
# Create hourly time intervals

df_clean["Time_hour_bin"] = (
    df_clean["Time"] // 3600
).astype(int)


# Calculate transactions and fraud cases per hour

hourly_analysis = df_clean.groupby(
    "Time_hour_bin"
).agg(
    total_transactions=("Class", "count"),
    fraud_transactions=("Class", "sum")
)


# Calculate fraud percentage

hourly_analysis["fraud_percentage"] = (
    hourly_analysis["fraud_transactions"]
    / hourly_analysis["total_transactions"]
) * 100


# Display results

print(hourly_analysis)
```

```python
# Step 1: Calculate quartiles

Q1 = df_clean["Amount"].quantile(0.25)

Q3 = df_clean["Amount"].quantile(0.75)


# Step 2: Calculate IQR

IQR = Q3 - Q1


# Step 3: Calculate outlier boundaries

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR


# Step 4: Identify outliers

outliers = df_clean[
    (df_clean["Amount"] < lower_bound) |
    (df_clean["Amount"] > upper_bound)
]


# Step 5: Display results

print("Q1:", Q1)

print("Q3:", Q3)

print("IQR:", IQR)

print("Lower Bound:", lower_bound)

print("Upper Bound:", upper_bound)

print("Number of Outliers:", len(outliers))

print(
    "Outlier Percentage:",
    len(outliers) / len(df_clean) * 100
)
```

```python
# Detect outliers in important PCA features

features_to_check = ["V14", "V17", "V12"]

for feature in features_to_check:

    Q1 = df_clean[feature].quantile(0.25)

    Q3 = df_clean[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[feature] < lower_bound) |
        (df_clean[feature] > upper_bound)
    ).sum()

    print(f"\nFeature: {feature}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Outlier Count:", outlier_count)
    print("Outlier Percentage:", outlier_count / len(df_clean) * 100)
```

```python
# STEP 14.1 - Feature Selection

# Columns created only for EDA
analysis_only_columns = [
    "Amount_log",
    "Time_hours",
    "Time_hour_bin"
]

# Create a separate modeling dataset
model_df = df_clean.drop(
    columns=analysis_only_columns,
    errors="ignore"
).copy()

# Separate input features and target
X = model_df.drop(columns=["Class"])

y = model_df["Class"]

# Display results
print("Model Dataset Shape:", model_df.shape)

print("Input Features Shape:", X.shape)

print("Target Shape:", y.shape)

print("\nInput Feature Names:")

print(X.columns.tolist())

print("\nTarget Distribution:")

print(y.value_counts())
```

```python
# STEP 14.2 - Feature Engineering

import numpy as np
import pandas as pd

# Create a separate copy of original input features
X_engineered = X.copy()

# Create logarithmic transaction amount feature
X_engineered["Amount_log"] = np.log1p(
    X_engineered["Amount"]
)

# Verify the new feature
print("Original X Shape:", X.shape)

print("Engineered X Shape:", X_engineered.shape)

print("\nNew Feature Preview:")

display(
    X_engineered[["Amount", "Amount_log"]].head(10)
)
```

```python
# Check the engineered feature

print("Missing Values:", X_engineered.isnull().sum().sum())

print(
    "Infinite Values:",
    np.isinf(X_engineered.select_dtypes(include="number")).sum().sum()
)

print(
    "\nAmount Log Statistics:"
)

print(X_engineered["Amount_log"].describe())
```

```python
from sklearn.preprocessing import StandardScaler

print("StandardScaler imported successfully!")
```

```python
from sklearn.model_selection import train_test_split

# Split original features
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dimensions
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
# Initialize scaler
scaler = StandardScaler()

# Fit ONLY on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data using training statistics
X_test_scaled = scaler.transform(X_test)

print("Training scaling completed!")
print("Testing scaling completed!")
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)

print("Scaled training dataset:")
display(X_train_scaled.head())
```

```python
# Check training feature means
print("Training Feature Means:")

print(X_train_scaled.mean().head(10).round(4))


# Check training feature standard deviations
print("\nTraining Feature Standard Deviations:")

print(X_train_scaled.std().head(10).round(4))
```

```python
print("Training Set Class Distribution:")
print(y_train.value_counts())

print("\nTraining Set Class Percentages:")
print((y_train.value_counts(normalize=True) * 100).round(4))
```

```python
# Experiment A: Baseline training data

X_train_baseline = X_train_scaled.copy()
y_train_baseline = y_train.copy()

print("Baseline Training Shape:", X_train_baseline.shape)
print("Baseline Target Shape:", y_train_baseline.shape)

print("\nBaseline Class Distribution:")
print(y_train_baseline.value_counts())
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))

print("Calculated Class Weights:")
print(class_weights)
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))

print("Calculated Class Weights:")
print(class_weights)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(
    sampling_strategy="auto",
    random_state=42,
    k_neighbors=5
)

print("SMOTE initialized successfully!")
```

```python
# Apply SMOTE ONLY to training data

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("SMOTE applied successfully!")
```

```python
print("Original Training Shape:", X_train_scaled.shape)
print("Resampled Training Shape:", X_train_smote.shape)

print("\nOriginal Target Shape:", y_train.shape)
print("Resampled Target Shape:", y_train_smote.shape)
```

```python
print("Class Distribution After SMOTE:")
print(y_train_smote.value_counts())

print("\nClass Percentages After SMOTE:")
print(
    (y_train_smote.value_counts(normalize=True) * 100).round(2)
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.countplot(x=y_train, ax=axes[0])
axes[0].set_title("Before SMOTE")
axes[0].set_xlabel("Transaction Class")
axes[0].set_ylabel("Number of Transactions")

sns.countplot(x=y_train_smote, ax=axes[1])
axes[1].set_title("After SMOTE")
axes[1].set_xlabel("Transaction Class")
axes[1].set_ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

print("Logistic Regression imported successfully!")
```

```python
# Experiment A: Baseline Logistic Regression
lr_baseline = LogisticRegression(
    max_iter=1000,
    random_state=42
)


# Experiment B: Class-weighted Logistic Regression
lr_weighted = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)


# Experiment C: SMOTE-based Logistic Regression
lr_smote = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print("All three Logistic Regression models configured successfully!")
```

```python
print("Baseline Model:")
print(lr_baseline)

print("\nClass-Weighted Model:")
print(lr_weighted)

print("\nSMOTE Model:")
print(lr_smote)
```

```python
# Train baseline Logistic Regression

lr_baseline.fit(
    X_train_baseline,
    y_train_baseline
)

print("Baseline Logistic Regression trained successfully!")
```

```python
print("Number of iterations:", lr_baseline.n_iter_)
print("Maximum iterations allowed:", lr_baseline.max_iter)
```

```python
print("Number of model coefficients:",
      lr_baseline.coef_.shape)

print("Model intercept:",
      lr_baseline.intercept_)
```

```python
# Generate baseline predictions

y_pred_baseline = lr_baseline.predict(X_test_scaled)

# Generate fraud probabilities
y_prob_baseline = lr_baseline.predict_proba(X_test_scaled)[:, 1]

print("Predictions generated successfully!")

print("\nPredicted Class Distribution:")
print(pd.Series(y_pred_baseline).value_counts())

print("\nFirst 10 Fraud Probabilities:")
print(y_prob_baseline[:10])
```

```python
from scipy.stats import ks_2samp

print("KS test imported successfully!")
```

```python
drift_results = []

for feature in reference_data.columns:

    statistic, p_value = ks_2samp(
        reference_data[feature],
        current_data[feature]
    )

    drift_results.append({
        "Feature": feature,
        "KS Statistic": statistic,
        "P-Value": p_value,
        "Drift Detected": p_value < 0.05
    })

drift_df = pd.DataFrame(drift_results)

drift_df = drift_df.sort_values(
    "P-Value",
    ascending=True
)

display(drift_df.round(6))
```

```python
total_features = len(drift_df)

drifted_features = drift_df["Drift Detected"].sum()

print("TOTAL FEATURES:", total_features)
print("DRIFTED FEATURES:", drifted_features)

print(
    "DRIFT PERCENTAGE:",
    round(drifted_features / total_features * 100, 2),
    "%"
)
```

```python
from statsmodels.stats.multitest import multipletests

# Apply Benjamini-Hochberg correction
reject, adjusted_pvalues, _, _ = multipletests(
    drift_df["P-Value"],
    alpha=0.05,
    method="fdr_bh"
)

drift_df["Adjusted P-Value"] = adjusted_pvalues
drift_df["Significant After Correction"] = reject

display(
    drift_df.sort_values(
        "Adjusted P-Value"
    ).round(6)
)
```

```python
# Scale reference and current datasets
reference_scaled = rf_monitor_scaler.transform(
    reference_data
)

current_scaled = rf_monitor_scaler.transform(
    current_data
)

# Generate fraud probabilities
reference_probabilities = (
    rf_monitor_model.predict_proba(reference_scaled)[:, 1]
)

current_probabilities = (
    rf_monitor_model.predict_proba(current_scaled)[:, 1]
)

print("Reference predictions:", len(reference_probabilities))
print("Current predictions:", len(current_probabilities))

print("\nReference probability mean:",
      reference_probabilities.mean())

print("Current probability mean:",
      current_probabilities.mean())
```

```python
# Preserve feature names after scaling

reference_scaled = pd.DataFrame(
    rf_monitor_scaler.transform(reference_data),
    columns=reference_data.columns,
    index=reference_data.index
)

current_scaled = pd.DataFrame(
    rf_monitor_scaler.transform(current_data),
    columns=current_data.columns,
    index=current_data.index
)

# Regenerate probabilities
reference_probabilities = (
    rf_monitor_model.predict_proba(reference_scaled)[:, 1]
)

current_probabilities = (
    rf_monitor_model.predict_proba(current_scaled)[:, 1]
)

print("Reference predictions:", len(reference_probabilities))
print("Current predictions:", len(current_probabilities))

print("Reference mean:", reference_probabilities.mean())
print("Current mean:", current_probabilities.mean())

print("Feature-name warning should now be resolved.")
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    reference_probabilities,
    bins=50,
    color="blue",
    label="Reference",
    stat="density",
    alpha=0.4
)

sns.histplot(
    current_probabilities,
    bins=50,
    color="orange",
    label="Current",
    stat="density",
    alpha=0.4
)

plt.title("SentinelAI - Prediction Distribution Comparison")
plt.xlabel("Predicted Fraud Probability")
plt.ylabel("Density")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
prediction_comparison = pd.DataFrame({
    "Reference": pd.Series(reference_probabilities).describe(),
    "Current": pd.Series(current_probabilities).describe()
})

display(prediction_comparison.round(6))
```

```python
reference_predictions = (
    reference_probabilities >= rf_monitor_threshold
).astype(int)

current_predictions = (
    current_probabilities >= rf_monitor_threshold
).astype(int)

reference_fraud_rate = (
    reference_predictions.mean() * 100
)

current_fraud_rate = (
    current_predictions.mean() * 100
)

print("REFERENCE PREDICTION FRAUD RATE:",
      round(reference_fraud_rate, 4), "%")

print("CURRENT PREDICTION FRAUD RATE:",
      round(current_fraud_rate, 4), "%")

print("REFERENCE FRAUD PREDICTIONS:",
      reference_predictions.sum())

print("CURRENT FRAUD PREDICTIONS:",
      current_predictions.sum())
```

```python
from scipy.stats import ks_2samp

prediction_ks_stat, prediction_p_value = ks_2samp(
    reference_probabilities,
    current_probabilities
)

print("PREDICTION DRIFT TEST")
print("---------------------")

print("KS Statistic:", round(prediction_ks_stat, 6))
print("P-Value:", round(prediction_p_value, 6))

if prediction_p_value < 0.05:
    print("Result: Statistical prediction drift detected")
else:
    print("Result: No statistically significant prediction drift")
```
